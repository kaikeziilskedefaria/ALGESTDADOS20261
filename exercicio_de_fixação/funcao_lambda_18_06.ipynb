```python
#01
#Lambda — Expressão de Identidade Funcional
#Em programação funcional, funções lambda são expressões anônimas de uso único. Crie uma função lambda que
#receba um número inteiro fornecido pelo usuário e retorne o seu quadrado. Armazene a lambda em
#uma variável chamada quadrado, aplique-a ao valor digitado e exiba o resultado com uma mensagem descritiva.

variavel = int(input(f"Digite um número inteiro "))

quadrado = lambda x: x * x
result = quadrado (variavel)
print("O quadrado é:", result )
```

```python
#02
#Lambda — Verificação de Paridade
#Lambdas permitem encapsular verificações lógicas de forma compacta. Crie uma função lambda chamada eh_par que
#receba um inteiro e retorne True se ele for divisível por 2 e False caso contrário.
#Solicite um número ao usuário, aplique a lambda e exiba a classificação par ou ímpar com base no retorno.

variavel = int(input(f"Digite um número inteiro "))

eh_par = lambda x: x % 2 == 0
result = eh_par (variavel)
if variavel:
    print(f"O número é par ")
else:
    print(f"O número é ímpar ")
```

```python
#03
#Map — Transformação em Massa de Dados
#A função map() aplica uma transformação a todos os elementos de um iterável sem laços explícitos.
#Solicite ao usuário 6 números inteiros e armazene-os em uma lista. Utilize map() com uma lambda
#para gerar uma nova lista com o dobro de cada valor. Exiba a lista original e a transformada lado a lado.

numeros = list(map(int, input("Digite 6 números inteiros separados por espaço: ").split()))

dobro = list(map(lambda x: x * 2, numeros))

print("Lista original:", numeros)
print("Lista com o dobro:", dobro)
```

```python
#04
#Map — Conversão de Unidades de Temperatura
#Em sistemas de monitoramento internacional, conversões de unidade são aplicadas a conjuntos de dados.
#Dado uma lista de temperaturas em Celsius fornecida pelo usuário, utilize map()
#com a fórmula F = C × 9/5 + 32 para converter toda a lista para Fahrenheit. Exiba os pares de conversão.

numeros = list(map(float, input("Digite a temperatura em Celsius separados por espaço: ").split()))

temp = list(map(lambda x: x * 9/5 + 32, numeros))

print("Lista Celsius:", numeros)
print("Lista com o Fahrenheit:", temp)
```
