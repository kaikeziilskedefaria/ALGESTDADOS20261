```python
#1 Validação de Limite Numérico
#Em um sistema de controle de acesso, é necessário verificar se um valor ultrapassa um limite definido. Elabore
#um programa em Python que solicite ao usuário a entrada de um número inteiro e determine se ele é maior que 10.
#Implemente a função verificar_maior_que_dez(numero), que receba um inteiro
#e retorne um valor booleano. O programa deve exibir uma mensagem descritiva ao usuário com base no resultado.

def verificar_maior_que_dez(numero):
    if numero > 10:
        return True
    else:
        return False

valor = int(input(f"Digte um número inteiro "))
if verificar_maior_que_dez(valor):
    print("O número é maior que 10")
else:
     print("O número não é maior que 10")
```

```python
#02
#Classificação de Paridade
#Em sistemas de validação de dados numéricos, identificar se um valor é par ou ímpar é uma operação fundamental.
#Desenvolva um programa em Python que leia um número inteiro fornecido pelo usuário e informe sua paridade.
#Crie a função eh_par(numero) que retorne True
#se o número for divisível por 2 e False caso contrário. O programa deve exibir o resultado de forma clara.

def eh_par(numero):
    if numero % 2 == 0:
        return True
    else:
        return False

numero = int(input(f"Digite um número inteiro"))
if eh_par(numero):
    print(f"O número é par")
else:
     print(f"O número é ímpar")
```

```python
#03
#Busca de Caractere em String
#Em um sistema de cadastro, regras de validação de nomes são aplicadas automaticamente.
#Escreva um programa que leia um nome informado pelo usuário e verifique se ele contém a letra "a"
# (maiúscula ou minúscula). Implemente a função
#contem_letra_a(nome) que retorne um valor booleano conforme a condição. Exiba uma mensagem adequada ao usuário.


def contem_letra_a(nome):
    return 'a' in nome.lower()


nome = input("Digite um nome: ")


if contem_letra_a(nome):
    print("O nome contém a letra 'a'.")
else:
    print("O nome NÃO contém a letra 'a'.")
```

```python
#04
#Verificação de Múltiplo de 5
#Em sistemas de análise numérica, a divisibilidade é um critério frequentemente utilizado. Desenvolva um programa
#que leia um número inteiro e informe se ele é múltiplo de 5. Crie a função eh_multiplo_de_5(numero) que retorne
#um valor lógico com base na divisibilidade por 5. O programa deve exibir o resultado de forma interpretativa.

def eh_multiplo_de_5(numero):
    if numero % 5 == 0:
        return True
    else:
        return False

numero = int(input(f"Digite um número inteiro: "))
if eh_multiplo_de_5(numero):
    print(f"O número é multiplo de 5")
else:
    print(f"O número não é multiplo de 5")
```

```python
#05
#Classificação de Temperatura Ambiente
#Em um sistema de monitoramento ambiental, as temperaturas precisam ser classificadas para acionar alertas.
#Escreva um programa que leia a temperatura atual em graus Celsius e informe se o clima está quente (acima de 30°C)
#ou frio (30°C ou menos).
#Implemente a função classificar_temperatura(temp) que retorne a classificação correspondente como string.

def classificar_temperatura(temp):
    if temp > 30:
        return "quente"
    else:
        return "frio"

temp = float(input("Digite a temperatura em Celsius: "))
print(f"A temperatura está {classificar_temperatura(temp)}!")
```

```python
# 06
# Controle de Fluxo Interativo (S/N)
# Em sistemas interativos, o usuário frequentemente precisa confirmar ou cancelar ações. Desenvolva um programa
#em Python que solicite ao usuário a entrada de um caractere: "S" para continuar ou "N" para sair. Crie a
#função validar_opcao(opcao) que retorne a mensagem adequada para cada caso, tratando também entradas inválidas.

def validar_opcao(opcao):
    if opcao == "S":
        return "Continuar"
    elif opcao == "N":
        return "Sair"
    else:
        return "Inválido"

opcao_digitada = input(f"Digite o caractere (S para Continuar, N para Sair): ").upper()

resultado_validacao = validar_opcao(opcao_digitada)

if resultado_validacao == "Continuar":
    print(f"Entrada permitida ! ")
elif resultado_validacao == "Sair":
    print(f"Saindo do programa. ")
else:
    print(f"Acesso negado ! ")
    print(f"Caractere inválido ! ")
```
