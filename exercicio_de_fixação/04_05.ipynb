```python
#Crie sua própria lista


#Crie uma lista chamada notas com 5 notas de 0 a 10. Depois: (a) imprima a primeira e a última nota; (b) altere a segunda nota para 10;
#(c) adicione uma nova nota com append(); (d) imprima o total de notas com len().

notas = [6,6.5,7,5,9]
print(f"{notas[0]} e {notas[-1]}")
notas[1] = 10
notas.append (9)
print(f"Quantidade de notas: {len(notas)}")
```

```python
#Removendo e organizando elementos da lista


#Crie uma lista chamada valores com pelo menos 6 números inteiros; em seguida, remova um
#elemento específico usando remove(), remova o último elemento com pop(), ordene a lista com
#sort() e, por fim, utilize um for para exibir cada valor da lista junto com sua posição (índice).

lista = ["9","2","3","1","5","6"]
lista.remove("3")
lista.sort()
lista.pop()
for i in range(len(lista)):
    print(lista)
```
