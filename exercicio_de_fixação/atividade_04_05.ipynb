```python
#01
#Contagem de Elementos
#Crie uma lista contendo os nomes de 5 vilões clássicos do Mario. Utilize a função len() para contar o número total
#de itens e exiba o resultado no terminal.

viloes = ["Bowser", "Wario", "Waluigi","King Boo","Kamek",]
print(f"A lista possui {len(viloes)} vilões")
```

```python
#02
#Adição Dinâmica
#Inicie uma lista vazia chamada moedas_coletadas. Use um laço de repetição para pedir que o usuário informe 4
#valores inteiros e adicione cada um deles ao final da lista usando append().

moedas_coletadas = [","]
inteiros = int(input("Informe 4 valores inteiros :"))
moedas_coletadas.append(inteiros)
print(moedas_coletadas)
```

```python
#03
#Inserção por Posição
#Crie a lista herois = ['Mario', 'Peach']. Utilize o método insert() para adicionar o nome 'Luigi' exatamente no
#meio da lista (índice 1) e exiba a lista final.

lista = ["Mario","Peach"]
lista.insert(1,"Luigi")
print(lista)
```

```python
#04
#Exclusão por Nome
#Dada a lista viloes = ['Bowser', 'Goomba', 'Wario', 'Koopa'], solicite que o usuário digite um nome. Se o nome
#estiver na lista, use remove() para excluí-lo.

viloes = ['Bowser', 'Goomba', 'Wario', 'Koopa']

nome = input("Digite um nome: ")

if nome in viloes:
    viloes.remove(nome)
```
