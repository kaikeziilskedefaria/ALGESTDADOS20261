```python
# ============================================
# 15 - Loja de Eletrônicos
# ============================================

preco = float(input('Digite o valor do produto (0 encerra): '))

total = 0

while preco != 0:
    if preco > 200:
        preco = preco - (preco * 0.05)

    total += preco

    preco = float(input('Digite o valor do produto (0 encerra): '))

print('Valor total a pagar: R$', round(total, 2))
```

```python
# ============================================
# 16 - Tempo de Atendimento
# ============================================

tempo = float(input('Tempo de atendimento (0 encerra): '))

clientes = 0
soma = 0

while tempo != 0:
    soma += tempo
    clientes += 1

    tempo = float(input('Tempo de atendimento (0 encerra): '))

if clientes > 0:
    media = soma / clientes
else:
    media = 0

print('Total de clientes atendidos:', clientes)
print('Tempo medio de atendimento:', round(media, 2))
```

```python
# ============================================
# 17 - Análise de 10 Números
# ============================================

negativos = 0
soma_positivos = 0

for i in range(10):

    num = int(input('Digite um numero: '))

    if i == 0:
        maior = num

    if num < 0:
        negativos += 1
    else:
        soma_positivos += num

    if num > maior:
        maior = num

print('Quantidade de negativos:', negativos)
print('Soma dos positivos:', soma_positivos)
print('Maior valor:', maior)
```

```python
# ============================================
# 18 - Consumo de Energia
# ============================================

total = 0

for i in range(6):

    consumo = float(input('Consumo da residencia: '))

    if consumo > 100:
        valor = consumo * 0.80
    else:
        valor = consumo * 0.50

    total += valor

print('Total arrecadado: R$', round(total, 2))
```

```python
# ============================================
# 19 - Adivinhe o Número Secreto
# ============================================

secreto = 7

tentativa = int(input('Digite um numero: '))

tentativas = 1

while tentativa != secreto:
    if tentativa > secreto:
        print('O numero secreto e menor')
    else:
        print('O numero secreto e maior')

    tentativa = int(input('Digite um numero: '))
    tentativas += 1

print('Numero correto!')
print('Quantidade de tentativas:', tentativas)
```

```python
# ============================================
# 20 - Estatísticas de Notas
# ============================================

nota = float(input('Digite uma nota (-1 encerra): '))

soma = 0
cont = 0
maior = 0

while nota != -1:
    soma += nota
    cont += 1

    if nota > maior:
        maior = nota

    nota = float(input('Digite uma nota (-1 encerra): '))

if cont > 0:
    media = soma / cont
else:
    media = 0

nota = float(input('Digite novamente as notas para verificar abaixo da media (-1 encerra): '))

abaixo = 0

while nota != -1:
    if nota < media:
        abaixo += 1

    nota = float(input('Digite novamente as notas para verificar abaixo da media (-1 encerra): '))

print('Media:', round(media, 2))
print('Maior nota:', maior)
print('Notas abaixo da media:', abaixo)
```

```python
# ============================================
# 21 - Pesquisa Eleitoral
# ============================================

voto = int(input('Digite o voto (0 encerra): '))

c1 = 0
c2 = 0
c3 = 0
invalidos = 0

while voto != 0:
    if voto == 1:
        c1 += 1
    elif voto == 2:
        c2 += 1
    elif voto == 3:
        c3 += 1
    else:
        invalidos += 1

    voto = int(input('Digite o voto (0 encerra): '))

validos = c1 + c2 + c3

if c1 > c2 and c1 > c3:
    vencedor = 1
elif c2 > c1 and c2 > c3:
    vencedor = 2
else:
    vencedor = 3

print('Votos validos:', validos)
print('Votos invalidos:', invalidos)
print('Candidato vencedor:', vencedor)
```

```python
# ============================================
# 22 - Estacionamento
# ============================================

tempo = float(input('Tempo de permanencia (0 encerra): '))

total = 0

while tempo != 0:
    valor = tempo * 10

    if tempo > 5:
        valor = valor - (valor * 0.10)

    total += valor

    tempo = float(input('Tempo de permanencia (0 encerra): '))

print('Total arrecadado: R$', round(total, 2))
```

```python
# ============================================
# 23 - Divisíveis por 2 e por 3
# ============================================

num = int(input('Digite um numero (0 encerra): '))

cont = 0
soma = 0
quantidade = 0

while num != 0:
    if num % 2 == 0 and num % 3 == 0:
        cont += 1

    soma += num
    quantidade += 1

    num = int(input('Digite um numero (0 encerra): '))

if quantidade > 0:
    media = soma / quantidade
else:
    media = 0

print('Quantidade divisivel por 2 e 3:', cont)
print('Media dos numeros:', round(media, 2))
```

```python
# ============================================
# 24 - Classificação Escolar
# ============================================

aprovado = 0
recuperacao = 0
reprovado = 0

for i in range(8):

    media = float(input('Digite a media do aluno: '))

    if media >= 7:
        aprovado += 1
    elif media >= 5:
        recuperacao += 1
    else:
        reprovado += 1

print('Aprovados:', aprovado)
print('Recuperacao:', recuperacao)
print('Reprovados:', reprovado)
```

```python
# ============================================
# 25 - Análise de Vendas
# ============================================

venda = float(input('Digite o valor da venda (0 encerra): '))

soma = 0
cont = 0

while venda != 0:
    soma += venda
    cont += 1

    venda = float(input('Digite o valor da venda (0 encerra): '))

if cont > 0:
    media = soma / cont
else:
    media = 0

venda = float(input('Digite novamente as vendas (0 encerra): '))

acima = 0

while venda != 0:
    if venda > media:
        acima += 1

    venda = float(input('Digite novamente as vendas (0 encerra): '))

print('Media das vendas:', round(media, 2))
print('Vendas acima da media:', acima)
```

```python
# ============================================
# 26 - Perfil Etário
# ============================================

idade = int(input('Digite a idade (0 encerra): '))

soma = 0
cont = 0
maiores = 0

while idade != 0:
    soma += idade
    cont += 1

    if idade >= 18:
        maiores += 1

    idade = int(input('Digite a idade (0 encerra): '))

if cont > 0:
    media = soma / cont
else:
    media = 0

print('Media das idades:', round(media, 2))
print('Maiores de idade:', maiores)
```

```python
# ============================================
# 27 - Par ou Ímpar até N
# ============================================

n = int(input('Digite o valor de N: '))

for i in range(1, n + 1):

    if i % 2 == 0:
        print(i, '- Par')
    else:
        print(i, '- Impar')
```

```python
# ============================================
# 28 - Caça aos Primos
# ============================================

num = int(input('Digite um numero (-1 encerra): '))

primos = 0
soma = 0

while num != -1:
    i = 1
    divisores = 0

    while i <= num:
        if num % i == 0:
            divisores += 1

        i += 1

    if divisores == 2:
        primos += 1
    else:
        soma += num

    num = int(input('Digite um numero (-1 encerra): '))

print('Quantidade de numeros primos:', primos)
print('Soma dos nao primos:', soma)
```

```python
# ============================================
# 29 - Avaliação de Notas
# ============================================

nota = float(input('Digite uma nota (0 encerra): '))

soma = 0
cont = 0

while nota != 0:
    soma += nota
    cont += 1

    nota = float(input('Digite uma nota (0 encerra): '))

if cont > 0:
    media = soma / cont
else:
    media = 0

nota = float(input('Digite novamente as notas (0 encerra): '))

acima = 0

while nota != 0:
    if nota > media:
        acima += 1

    nota = float(input('Digite novamente as notas (0 encerra): '))

print('Media geral:', round(media, 2))
print('Notas acima da media:', acima)
```

```python
# ============================================
# 30 - Cálculo de Fatorial
# ============================================

n = int(input('Digite um numero inteiro positivo: '))

fat = 1

for i in range(1, n + 1):
    fat = fat * i

print('Fatorial:', fat)

if fat % 2 == 0:
    print('O fatorial e par')
else:
    print('O fatorial e impar')
```
